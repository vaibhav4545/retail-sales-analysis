**Step 3 final: create the flag**

```python
non_merch_codes = ['EB', 'COTHERS', 'DEFAULT', 'EARN', 'CTRIMS', 'CBG', 'BCP']
```

```python
df_clean['is_non_merch'] = df_clean['type_code'].isin(non_merch_codes)
```

```python
df_clean['is_non_merch'].sum()
```

```python
df_clean.groupby('is_non_merch')['line_item_amount'].agg(['count', 'sum', 'mean']).round(0)
```

```python
df_clean.shape
```

```python
orig_cols = ['user_id', 'bill_id', 'line_item_amount', 'bill_discount', 'transaction_date', 'description', 'inventory_category', 'colour', 'size', 'zone_name', 'store_name', 'year']
```

```python
df_clean['is_repeat_row'] = df_clean.duplicated(subset=orig_cols)
```

```python
df_clean['is_repeat_row'].sum()
```

```python
df_clean[~df_clean['is_non_merch']].groupby('zone_name')['is_repeat_row'].mean().round(3)
```

```python
df_clean.shape
```

```python
df_clean[~df_clean['is_non_merch']]['is_repeat_row'].sum()
```

```python
df_clean['bill_total'] = df_clean.groupby('bill_id')['line_item_amount'].transform('sum')
```

```python
df_clean.drop_duplicates('bill_id')['bill_total'].quantile([0.5, 0.9, 0.99, 0.999]).round(0)
```

```python
df_clean.shape
```

```python
bulk_threshold = df_clean.drop_duplicates('bill_id')['bill_total'].quantile(0.99)
bulk_threshold
```

```python
df_clean['is_bulk_bill'] = df_clean['bill_total'] > bulk_threshold
```

```python
df_clean['is_bulk_bill'].sum()
```

```python
df_clean.groupby('is_bulk_bill')['line_item_amount'].agg(['count', 'sum']).round(0)
```
